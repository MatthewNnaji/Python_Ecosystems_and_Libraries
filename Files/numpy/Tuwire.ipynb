## Welcome to the absolute beginner’s guide to NumPy!
NumPy (Numerical Python) is an open source Python library that’s widely used in science and engineering. The NumPy library contains multidimensional array data structures, such as the homogeneous, N-dimensional ndarray, and a large library of functions that operate efficiently on these data structures. Starting basic array work!!! - 3 x 3 matrix (or 2dim array) - shape, dimensions, slicing

```python
faraday =np.array([[1,2,3],[3,6,9],[9,18,27]])
```

```python
faraday[1:] #calls everything but the first row
```

```python
faraday[:1] #calls nothing but the first row
```

```python
faraday[1:,1:] #first row
```

### [2] More Complex Array Functions [arange,linspace,full]
-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-=-


**Array range (arange) and linear space (linspace)**

```python
np.arange(2,20,5) #np.arange(first number, last number, step) 
```

```python
np.linspace(0, 375, num=4) #first number 0, last nunber 375; creates an array where each point is equidistant 
#with these paramaters - each point has 125 between them
```

as you see here; 375/3 = 125 - but since starting at zero, add one to get the desired linear space.
for instance, 375/15 = 25 and 375/75 = 5; for both we would need to add one (see below)

```python
print(np.linspace(0,375,num=16),"<- first | second ->",np.linspace(25,375,num=15))
```

```python
np.linspace(0,375,num=76)
```

```python
print(np.linspace((10,20),(990,1000),50)) #two dimensional linspaces
```
