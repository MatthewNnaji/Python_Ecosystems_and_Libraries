## [2a] Series in pandas
introducing series and basic locating techniques

```python
line = ['Bulbasaur','Ivysaur','Venusaur','MegaVenusaur']
series = pd.Series(line)
series
```

***i) we can change the index to make it more readable - whether its a tuple, list, a np.array etc.***

```python
print(pd.Series(line, index=[1,2,3,3]))
print('')
names1= ['Stage1 |','Stage2 |','Stage3 |','Mega   |']
nametest =pd.Series(line,index=names1)
nametest
```

***ii) we can print a specific value, i.e. when a label corresponds to a value using the series.loc[] function:***

```python
print(nametest.loc["Stage3 |"],nametest.loc["Stage1 |"])
```

***iii) replacing a specific value using location:***

```python
nametest.loc["Mega   |"] = 'GmaxVenusaur'
print(series)
print(nametest)
```

*iv) this can be done via index with iloc (index location)*

```python
print(nametest.iloc[3])
```

## [2b] Filtering series
basic filtering.

***i) start with a new data set. we will use this to do some filtering***

```python
data = [66,88,200,200.5,100,180,10,300,202,400,800]
alphanumeric='abcdefghijklmnopqrstuvwxyz'

stringr = []

for i in alphanumeric:
    stringr+=i

numSeries =pd.Series(data,index=stringr[:len(data)])  #creates a sublist of the alphabet for the index, only up to the length of the data
numSeries
```

***ii) now create a filtered sub-series, for values greater than or equal to 200***

```python
subnumero = numSeries[numSeries >=200]
subnumero
```

***iii) now we will explore pandas with dictionaries :)***

```python
calories = {"Day1":1750, "Day2":2100, "Day3":1700}
pizza = pd.Series(calories)
pizza #notice how the labels are automatically used as a column!!!
```

```python
print(pizza.loc['Day3'])
pizza.iloc[2] = 3800 #updating again, this time with iloc
pizza
```

```python
print(pizza[pizza<2000]) #all the days i kept my diet :))
```

## [7] Pivot Tables
done in both possible ways, an explicit functions or R-style

*pandas.pivot_table(data, values=None, index=None, columns=None, aggfunc='mean', fill_value=None, margins=False, dropna=True, margins_name='All', observed=True, sort=True,* ***kwargs)*


```python
frequency = gen1.groupby('Type1')
frequency = frequency['Name'].count()
print(frequency)

frequency.info()
widetypes1 = pd.pivot_table(gen1,values=['Height','Weight'],columns='Type1',aggfunc='mean')
widetypes1.loc['Index'] = list(range(1,16))
widetypes1.loc['Frequency'] = frequency

widetypes1.iloc[3] = widetypes1.iloc[3].astype(int)
```

```python
widetypes1
```

```python
widetypes1.insert(6,'Dark',["NA","NA",7.5,0])
widetypes1.insert(16, "Steel", ['NA','NA',17,0])
widetypes1.insert(16, "Flying", [1.626316,61.247368,18,19])
widetypes1.loc['Index'] = list((range(1,19)))
widetypes1
```

```python
print(pd.pivot_table(gen1,values=['Height','Weight'],columns='Type2',aggfunc='mean'))
chuckit = gen1.groupby('Type2')
print(chuckit['Name'].count())
```

```python
widetypes2 = pd.pivot_table(gen1,values=['Height','Weight'],columns='Type2',aggfunc='mean')
widetypes2.loc['Frequency'] = chuckit['Name'].count()
widetypes2
```
